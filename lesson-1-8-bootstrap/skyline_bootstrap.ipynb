## Part A. Bootstrap CI for the mean.

```python
import pandas as pd
import numpy as np

skyline = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")
data = skyline["hours_studied"].values

n = len(data)
B= 10000

np.random.seed(42)

sample_mean = data.mean()
sample_std = data.std()

#standard error of the mean
standard_error= sample_std / np.sqrt(n)

#critical value for 95% confidence level
z= 1.96

#margin of error
moe = z * standard_error

#Classical 95% CI
ci_lower = sample_mean - moe
ci_upper = sample_mean + moe

#Bootstrap 95% CI
boot_means = np.array([
    np.random.choice(data, size=n, replace =True).mean()
    for _ in range(B)
])

ci_mean = np.percentile(boot_means, [2.5, 97.5])



print(f"Sample mean: {sample_mean:.2f}")
print(f"Classical 95% CI: ({ci_lower:.2f}, {ci_upper:.2f})")
print(f"Bootstrap 95% CI mean: ({ci_mean[0]:.2f}, {ci_mean[1]:.2f})")

```

The classical and bootstrap confidence intervals are very similar because hours studied is approximately normally distributed. Since the normality assumption is reasonable, the classical method works well. Bootstrap provides a similar result, confirming the estimate.

## Part B. Bootstrap CI for the median.

```python
sample_median = np.median(data)

#Bootstrap 95% CI
boot_medians = np.array([
    np.median(np.random.choice(data, size =n, replace = True))
    for _ in range(B)
])

ci_median = np.percentile(boot_medians, [2.5, 97.5])


print(f"Sample median: {sample_median:.2f}")
print(f"Bootstrap 95% CI median: ({ci_median[0]:.2f}, {ci_median[1]:.2f})")
```

The 95% confidence interval shows the range of plausible values for the population median, with a narrower interval indicating that the median is estimated more precisely.

```python
## Part C. Bootstrap CI for the 95th percentile.

sample_95th = np.percentile(data, 95)

boot_percentiles = np.array([
    np.percentile(np.random.choice(data, size =n, replace = True), 95)
    for _ in range(B)
])

ci_percentile = np.percentile(boot_percentiles, [2.5, 97.5])


print(f"Sample 95th percentile: {sample_95th:.2f}")
print(f"Bootstrap CI for 95% : ({ci_percentile[0]:.2f}, {ci_percentile[1]:.2f})")
```

The 95th percentile CI tends to be wider than the median CI because the median is estimated from the middle of the data, where there are many observations around it. The 95th percentile is estimated from the extreme upper end of the data, where there are fewer observations, so the estimate is more uncertain and its confidence interval is wider.

```python
import plotly.express as px

bootstrap_df = pd.DataFrame({
    "boot_mean": boot_means,
    "boot_median": boot_medians,

})

fig_mean = px.histogram(
    bootstrap_df,
    x="boot_mean",
    nbins=100,
    title="Bootstrap Distribution: Mean Hours studied",

)
fig_mean.add_vline(x=ci_mean[0], line_dash="dash", annotation_text="2.5%")
fig_mean.add_vline(x=ci_mean[1], line_dash="dash", annotation_text="97.5%")
fig_mean.show()

fig_median = px.histogram(
    bootstrap_df,
    x="boot_median",
    nbins=100,
    title="Bootstrap Distribution: Median Hours studied",
)
fig_median.add_vline(x=ci_median[0], line_dash="dash", annotation_text="2.5%")
fig_median.add_vline(x=ci_median[1], line_dash="dash", annotation_text="97.5%")
fig_median.show()
```
