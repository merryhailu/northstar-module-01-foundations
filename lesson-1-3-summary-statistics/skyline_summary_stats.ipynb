# Nominal columns: count, unique count, mode.

```python
enrollment_counts = skyline['enrollment_id'].count()
enrollment_counts_unique = skyline['enrollment_id'].nunique()
enrollment_mode = skyline['enrollment_id'].mode()

print(f"Total enrollments: {enrollment_counts}")
print(f"Unique enrollments: {enrollment_counts_unique}")
print(f"Mode of enrollments: {enrollment_mode.values[0]}")
```

```python
course_counts = skyline['course_name'].count()
course_unique_counts = skyline['course_name'].nunique()
course_mode = skyline['course_name'].mode()

print(f"Total courses: {course_counts}")
print(f"Unique courses: {course_unique_counts}")
print(f"Mode of courses: {course_mode.values[0]}")
```

```python
completion_counts = skyline['completion_status'].count()
completion_unique_counts = skyline['completion_status'].nunique()
completion_mode = skyline['completion_status'].mode()

print(f"Total completion statuses: {completion_counts}")
print(f"Unique completion statuses: {completion_unique_counts}")
print(f"Mode of completion statuses: {completion_mode.values[0]}")
```

# Ordinal columns: mode, median, count.

```python
final_grade_counts = skyline['final_grade'].count()
final_grade_unique_counts = skyline['final_grade'].nunique()
final_grade_mode = skyline['final_grade'].mode()


print(f"Total final grades: {final_grade_counts}")
print(f"Unique final grades: {final_grade_unique_counts}")
print(f"Mode of final grades: {final_grade_mode.values}")
```

### Mode and Mean of final grade


The mode is bimodal, with **B and C** having the highest frequencies because we gave them higher probabilities when we generated the data.

Since the final grades are categorical string values (A, B, C, D, F), calculating the mean is not meaningful in the usual way. If we report a numerical mean by assigning numbers to the grades, we would need to include a **caveat** explaining how the grades were converted to numbers and that the resulting mean does not have a direct interpretation as an actual grade.

Therefore, the mode is more appropriate for reporting the most common final grades to a stakeholder.

#  Interval columns: mean, median, mode, standard deviation, IQR.

```python
enrollment_year_mean = skyline['enrollment_year'].mean()
enrollment_year_median = skyline['enrollment_year'].median()
enrollment_year_mode = skyline['enrollment_year'].mode()
enrollment_year_std = skyline['enrollment_year'].std()
enrollment_year_q1 = skyline['enrollment_year'].quantile(0.25)
enrollment_year_q3 = skyline['enrollment_year'].quantile(0.75)
enrollment_year_iqr = enrollment_year_q3 - enrollment_year_q1


print(f"Mean Enrollment year: {enrollment_year_mean:.2f}")
print(f"Median Enrollment year: {enrollment_year_median}")
print(f"Mode Enrollment year: {enrollment_year_mode.values}")
print(f"Standard Deviation Enrollment year: {enrollment_year_std:.2f}")
print(f"IQR Enrollment year: {enrollment_year_iqr}")

```

# Ratio columns: mean, median, mode, standard deviation, IQR, p95.

```python
hours_studied_mean = skyline['hours_studied'].mean()
hours_studied_median = skyline['hours_studied'].median()
hours_studied_mode = skyline['hours_studied'].mode()
hours_studied_std = skyline['hours_studied'].std()
hours_studied_q1 = skyline['hours_studied'].quantile(0.25)
hours_studied_q3 = skyline['hours_studied'].quantile(0.75)
hours_studied_iqr = hours_studied_q3 - hours_studied_q1
hours_studied_p95 = skyline['hours_studied'].quantile(0.95)

print(f"Mean Hours Studied: {hours_studied_mean:.2f}")
print(f"Median Hours Studied: {hours_studied_median:.2f}")
print(f"Mode Hours Studied: {hours_studied_mode.values}")
print(f"Standard Deviation Hours Studied: {hours_studied_std:.2f}")
print(f"IQR Hours Studied: {hours_studied_iqr:.2f}")
print(f"95th Percentile Hours Studied: {hours_studied_p95:.2f}")
```

### Mean and Median of Hours studied

As we can see above, the mean and median are close to each other. Since we used a normal distribution when we generated the data, this shows that the distribution is approximately symmetric. Because there are no significant outliers, the mean can be used as a measure of central tendency.
